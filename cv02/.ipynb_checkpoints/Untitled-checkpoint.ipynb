```python
import matplotlib.pyplot as plt
import plotly.express as px

data = [1, 2, 3]

# 1. Matplotlib
plt.figure(figsize=(6, 4))
plt.bar(range(len(data)), data)
plt.title("Sloupcový graf (Matplotlib)")
plt.xlabel("Index")
plt.ylabel("Hodnota")
plt.show()

# 2. Plotly
fig = px.bar(x=[0, 1, 2], y=data, labels={'x': 'Index', 'y': 'Hodnota'}, title="Sloupcový graf (Plotly)")
fig.show()
```
